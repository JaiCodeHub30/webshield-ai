# WebShield AI - EDA (Phase 1)
Class distribution of the cleaned Kaggle dataset and the domain-grouped train/test split. URLs are treated as strings only.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
P = Path('..') / 'data' / 'processed'
LABELS = ['benign', 'phishing', 'malware', 'defacement']
df = pd.read_csv(P / 'clean.csv'); train = pd.read_csv(P / 'train.csv'); test = pd.read_csv(P / 'test.csv')
print(len(df), len(train), len(test))
```

```python
counts = df['type'].value_counts().reindex(LABELS)
ax = counts.plot.bar(color='#4C72B0', figsize=(6, 4), title='Class distribution (clean data)')
ax.set_ylabel('URLs')
for i, v in enumerate(counts): ax.text(i, v, f'{v/counts.sum():.1%}', ha='center', va='bottom')
plt.tight_layout(); plt.show()
```

```python
split = pd.DataFrame({'train': train['type'].value_counts(normalize=True), 'test': test['type'].value_counts(normalize=True)}).reindex(LABELS)
split.plot.bar(figsize=(6, 4), title='Class share: train vs test'); plt.ylabel('share'); plt.tight_layout(); plt.show()
```

```python
overlap = set(train['domain']) & set(test['domain'])
print('domains in both train and test:', len(overlap))
assert not overlap
```

```python
df['url_len'] = df['url'].str.len()
df.boxplot(column='url_len', by='type', showfliers=False, figsize=(6, 4)); plt.suptitle(''); plt.title('URL length by class'); plt.show()
df.groupby('domain').size().sort_values(ascending=False).head(15)
```
